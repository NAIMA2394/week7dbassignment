```python
import pandas as pd
from sklearn.datasets import load_iris

# Load Iris dataset
iris = load_iris()

# Convert to pandas DataFrame
df = pd.DataFrame(iris.data, columns=iris.feature_names)

# Add the target (species) to the DataFrame
df['species'] = iris.target

# Show the first few rows of the dataset
df.head()
```

```python
# Inspect data types
df.dtypes

# Check for missing values
df.isnull().sum()
```

```python
# Compute basic statistics
df.describe()
```

```python
# Group by species and compute the mean for each group
df.groupby('species').mean()
```

```python
import matplotlib.pyplot as plt

# Calculate the mean for each feature per species
mean_values = df.groupby('species').mean()

# Plot a line chart for each feature
mean_values.plot(kind='line', marker='o')
plt.title('Mean Features per Species')
plt.ylabel('Feature Value')
plt.xlabel('Species')
plt.show()
```

```python
# Plot a bar chart for average sepal length per species
df.groupby('species')['sepal length (cm)'].mean().plot(kind='bar', color='skyblue')
plt.title('Average Sepal Length per Species')
plt.ylabel('Sepal Length (cm)')
plt.xlabel('Species')
plt.show()
```

```python
# Plot a histogram for petal length
df['petal length (cm)'].plot(kind='hist', bins=20, color='green', edgecolor='black')
plt.title('Petal Length Distribution')
plt.xlabel('Petal Length (cm)')
plt.show()
```

```python
# Scatter plot between sepal length and petal length
plt.scatter(df['sepal length (cm)'], df['petal length (cm)'], c=df['species'], cmap='viridis')
plt.title('Sepal Length vs Petal Length')
plt.xlabel('Sepal Length (cm)')
plt.ylabel('Petal Length (cm)')
plt.colorbar(label='Species')
plt.show()
```
